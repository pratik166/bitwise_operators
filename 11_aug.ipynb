## Assignment Operator : ->
### used to store a value or the result of an expression into a variable
### =

```python
a = 10 
a = a + 10 
print(a)
```

## Composite Assignment:->
## a shortcut that performs a math or bitwise operation on a variable and assigns the new result back to that same variable in one step
## +=, -=, *=, /=, //=, %=

```python
a = 10 
a = a + 10
print(a)
a += 5 
print(a)
```

```python
a = 20 
a -= 5
print(a)
```

```python
a = 10 
a *= 3
print(a)
```

```python
a = 30 
a /= 6
print(a)
```

```python
a = 20 
a //= 3
print(a)
```

```python
a = 50 
a %= 7
print(a)
```

## Bitwise Operators :->
### used to perform operations directly on the binary representations (bits) of integers. Python converts the decimal integers into binary digits, runs the logic bit-by-bit, and outputs the final result back in decimal format.
## &, |, ^, ~, >>, <<

## & (and) operator :->
#### Perform AND opreation on binary digits.
#### Compares corresponding bits; evaluates to 1 only if both matching bits are 1.

```python
print(5&3)
```

```python
print(10&5)
```

```python
print(7&3)
```

```python
print(34&12)
```

## | (OR) Operator :->
### Evaluates to 1 if either or both of the compared bits are 1.

```python
print(5|3)
```

```python
print(10|4)
```

```python
print(20|5)
```

```python
print(3|7)
```

```python
print(7|9)
```

## ^ (XOR) Operator :->
### Evaluates to 1 if the bits are different (1 and 0). It evaluates to 0 if they are identical (0 and 0, or 1 and 1).

```python
print(3^7)
```

```python
print(3^8)
```

```python
print(1^7)
```

```python
print(5^34)
```

```python
print(9^4)
```

## ~ (NOT) Operator :->
### Flips all bits (0 becomes 1 and vice versa).

```python
a = 6
print(~a)
```

```python
a = 9
print(~a)
```

```python
a = 3
print(~a)
```

```python
a = 78
print(~a)
```

```python
a = 23
print(~a)
```

## >> (Right shift) Operator :->
### Pushes bits down to the right side, dropping any values that fall off. 
### Mathematically, x >> n acts as floor division x // 2^n

```python
print(5>>2)
```

```python
print(8>>2)
```

```python
print(6>>4)
```

```python
print(23>>9)
```

```python
print(4>>5)
```

## << (Left shift) Operator :->
###  Pushes zeros into the right side of the bit pattern. 
### Mathematically, x << n acts as  x*2^n.


```python
print(8<<2)
```

```python
print(9<<7)
```

```python
print(5<<4)
```

```python
print(34<<5)
```

```python
print(34<<8)
```
